```python
import torchaudio
from speechbrain.inference.speaker import EncoderClassifier
classifier = EncoderClassifier.from_hparams(source="speechbrain/spkrec-ecapa-voxceleb")
```

```python
signal, fs = torchaudio.load('../assets/vi.wav')
reference = classifier.encode_batch(signal).squeeze(0)
```

```python
signal, fs = torchaudio.load('../assets/en_sample.mp3')
proccessed = classifier.encode_batch(signal).squeeze(0)
```

```python
import torch
torch.cosine_similarity(reference, proccessed)
```
